## String Indexing and slicing

**Positive Indexing:**
    
    Default values
    
    start_index: 0
    stop index: len(string)-1
    step size: 1

**Negative Indexing**

    Default values

    start_index: -len(string)
    stop index: -1
    step size:1

```python
# name = "Thaman Deshpande"
# print(name)
name = "Thaman Deshpande"
print(name)
```

**len(string_name) : returns the number of characters present in the string(also counts whitespaces, special characters)**

```python
# Check the length of the string 
# len(name)
len(name)
```

**slicing:**

variable_name[start_index:stop_index:step_size]

    stop_index : python internally subtracts 1 from this index
    always add 1 and then pass it 

    Exmaple: Des
    start index: 7
    stop index: 9+1 = 10
    step size: 1


```python
# Fetch data from 7th until 9th position where 9th position data is included(provide 9+1=10)
# name[7:10]
name[7:10]
```

```python
print(name[7:10:1])
```

```python
# start index: 3
# stop index: 11
# step size: 1
print(name[3:11])
print(name[3:11:1])
```

## String methods

1) index(character/substring): returns the position of character in the string. If character is not found, it will throw an error
2) find(character/substring): returns the position of character in the string. If character is not found, it will return -1
3) count(character/substring):returns the number of occurrences of character in the string
4) lower(): converts the string to lowercase
5) upper(): converts the string to uppercase
6) title(): converts the 1st character of every word in the string to uppercase and remaining all characters to lowercase
7) capitalize(): converts the 1st character of 1st word in the string to uppercase and remaining all characters to lowercase
8) replace(oldcharacter,newcharacter): replaces old data with new data in new copy of string
